```python
import pandas as pd

df = pd.read_csv("mavenclad_shortlisted_phi.csv")
tdf = pd.read_csv("nonzerotables.csv")

tables = tdf['table_name'].to_list()
df.shape, len(tables)
```

```python
df['TABLE_NAME'].nunique()
```

```python
df = df[df['TABLE_NAME'].isin(tables)]
df.shape
```

```python
df['TABLE_NAME'].nunique()
```

```python
completed = pd.read_csv("completed.csv")
ctables = completed['TABLE_NAME'].to_list()
len(ctables)
```

```python
df = df[~df['TABLE_NAME'].isin(ctables)]
df.shape
```

```python
df['TABLE_NAME'].nunique()
```

```python
patient_keys = {'patientid', 'personid', 'pid'}
encounter_keys = {'visitid', 'encounterid', 'encid'}

def normalize_colname(col: str) -> str:
    """Lowercase and strip non-alphanumeric characters so 'encId' -> 'encid'."""
    return ''.join(ch for ch in col.lower() if ch.isalnum())

# normalize first
df['normalized_col'] = df['COLUMN_NAME'].apply(normalize_colname)

# assign rules
df.loc[df['normalized_col'].isin(patient_keys), 'DE_IDENTIFICATION_RULE'] = 'PATIENT_ID'
df.loc[df['normalized_col'].isin(encounter_keys), 'DE_IDENTIFICATION_RULE'] = 'ENCOUNTER_ID'

# drop helper column if not needed
df.drop(columns=['normalized_col'], inplace=True)
```
