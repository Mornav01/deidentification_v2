```python
import re

file_path = '/Users/ndaidcnd/Desktop/deidentification/CDC/MySQL/cdc_merger.log' # Replace with your actual filename
failed_tables = set()

# Regex captures the text between "Merge failed for " and the colon
pattern = r"Merge failed for (.*?): .*?Connection refused"

try:
    with open(file_path, 'r') as file:
        for line in file:
            match = re.search(pattern, line)
            if match:
                failed_tables.add(match.group(1))

    print(f"Found {len(failed_tables)} failed tables:")
except FileNotFoundError:
    print("Error: The log file was not found.")
```
